데이터셋 정보

이 데이터 세트는 오듀본 협회 북미 버섯 야외 안내서(1981)에서 발췌한 양송이과(Agaricus)와 덩굴버섯과(Lepiota)에 속하는 23종의 아가미버섯에 대한 가상 표본 설명을 포함합니다. 각 종은 확실히 식용 가능, 확실히 독성, 또는 식용 가능 여부가 불확실하고 권장하지 않는 것으로 분류됩니다. 마지막 분류는 독성 분류와 통합되었습니다. 안내서에는 버섯의 식용 여부를 판단하는 간단한 규칙이 없으며, 옻나무나 담쟁이덩굴처럼 "잎이 세 개면 먹지 마라"와 같은 규칙은 없다고 명시되어 있습니다.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('mushrooms.csv')
df.head(5) 
```

```python
# 데이터 타입과 결측치 확인
df.describe()
print(df.isna().sum().sum())   # NaN 개수 합계
print((df == '?').sum()[(df == '?').sum() > 0])   # '?'로 숨은 결측
```

23개 칼럼 모두 object(문자) 타입 → 모델에 넣으려면 인코딩 필요.
NaN은 0개지만, stalk-root에 '?'로 표시된 결측이 2,480개 숨어 있음.

```python
print(df.shape)
print(df['class'].value_counts())
print(df['stalk-root'].value_counts())
```

식용(e) 4208 개 / 독성(p) 3916개 - 거의 반반임.

앞서 칼럼 정보에서 'stalk-root' 혼자 ?(결측치)를 가지고 있었으므로 value_counts()로 확인. 총 2480개의 누락 확인.

```python
# stalk-root(줄기 뿌리 모양) 과의 관계 확인하기
#pd.crosstab(df['stalk-root'], df['class'], normalize='index')
df.groupby('stalk-root')['class'].value_counts(normalize=True)
```

groupby()로 그룹별 class 비율 확인

stalk-root 값(b, c, e, r, ?)을 행으로 class값(e, p)을 열로 놓고
각 조합이 몇 번 나왔는지 알려주는 것이다.

독성(p) 비율이 71%로, 전체 평균보다 높다.
결측 자체가 독성 신호라는 뜻.
결측값을 처리X

```python
# nunique()로 컬럼마다 고유값이 몇 개인지 확인.
print(df.nunique())
```

veil-type(막 유형) - 고유값이 1개뿐이라 모든 행이 같은 값이라는 뜻/
분류에 아무 정보를 주지 못함.

```python
# veil-type 삭제.
df = df.drop('veil-type', axis=1)
```

```python
# 레이블 인코딩
# 대상 : class (e/p)
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
# 어떤 값들이 있는지 학습
encoder.fit(df['class']) 
# 학습한 기준대로 변환 (e-> 0, p->1)
df['class'] = encoder.transform(df['class']) 


print(encoder.classes_)

```

```python
# 피처 원-핫 인코딩
# 대상 : 나머지 21개 컬럼(갓 모양, 냄새, 색깔 등)
# 21개 칼럼은 원-핫 인코딩 진행하는 이유는 숫자의 가중치에 따른 문제를
# 해결하기 위함.
# 모든 칼럼의 dtype이 object 였으므로 진행.

X = pd.get_dummies(df.drop('class', axis=1), dtype=int)
y = df['class']

print('인코딩 전:', df.drop('class', axis=1).shape)
print('인코딩 후:', X.shape)
```

인코딩 후 칼럼 수가 21개 칼럼 각각의 고유값 개수 만큼 늘어남.
21 -> 116

```python
# 전처리 후 새 DataFrame 저장

# 8124개 모두 사용했으나 필요 여부에 따라 샘플링을 할지 말지 결정해야함.

new_data = pd.concat([X, y], axis=1)
new_data.to_csv('mushrooms_preprocessed.csv', index=False)

# 원본 범주명 인코딩 전 파일
df_raw = pd.read_csv('mushrooms.csv').drop('veil-type', axis=1)
df_raw.to_csv('mushrooms_cleaned.csv', index=False)

```
